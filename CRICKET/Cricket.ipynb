```python
import pandas as pd

# Load the datasets
batting_df = pd.read_csv('Batting.csv')
bowling_df = pd.read_csv('Bowling.csv')
match_results_df = pd.read_excel('MatchResults-refined.xlsx')

# Display previews
print("=== Batting Dataset Preview ===")
display(batting_df.head())

print("\n=== Bowling Dataset Preview ===")
display(bowling_df.head())

print("\n=== Match Results Dataset Preview ===")
display(match_results_df.head())
```

### Question 1: Who are the top 10 batsmen based on their total runs scored across all matches?

```python
# Calculate top 10 batsmen based on total runs (using 'R' for Runs)
top_10_batsmen = batting_df.groupby('Player')['R'].sum().reset_index()
top_10_batsmen = top_10_batsmen.sort_values(by='R', ascending=False).head(10)
top_10_batsmen.columns = ['Player', 'Total Runs']
display(top_10_batsmen)
```

### Question 2: Which 10 players have the highest strike rate among players who have faced at least 50 balls?

```python
# Aggregate runs ('R') and balls ('B') faced
player_stats = batting_df.groupby('Player').agg({'R': 'sum', 'B': 'sum'}).reset_index()

# Filter for players who have faced at least 50 balls
players_min_50_balls = player_stats[player_stats['B'] >= 50].copy()

# Calculate overall strike rate
players_min_50_balls['Strike_Rate'] = (players_min_50_balls['R'] / players_min_50_balls['B']) * 100
top_10_sr = players_min_50_balls.sort_values(by='Strike_Rate', ascending=False).head(10)
display(top_10_sr[['Player', 'R', 'B', 'Strike_Rate']])
```

### Question 3: Which player has scored the most fours and which player has scored the most sixes?

```python
# Most fours
most_fours = batting_df.groupby('Player')['4s'].sum().idxmax()
max_fours_val = batting_df.groupby('Player')['4s'].sum().max()

# Most sixes
most_sixes = batting_df.groupby('Player')['6s'].sum().idxmax()
max_sixes_val = batting_df.groupby('Player')['6s'].sum().max()

print(f"Player with most fours: {most_fours} ({max_fours_val} Fours)")
print(f"Player with most sixes: {most_sixes} ({max_sixes_val} Sixes)")
```

### Question 4: Which team has won the highest number of matches, and how many matches did they win?

```python
# Since match_results_df might contain match-level data, let's aggregate wins.
# If it contains team-level or match-level info, we verify unique entries or sum the 'Win' field.
# Using unique Match boundaries if we grouping by 'Team' and 'Date' with 'Win' == 1
winning_teams = match_results_df[match_results_df['Win'] == 1].groupby('Team')['Date'].nunique().reset_index(name='Wins')
most_wins = winning_teams.sort_values(by='Wins', ascending=False).head(1)
display(most_wins)
```

### Question 5: For each match, who was the highest run scorer?

```python
# Group by Date, Team, Vs and find the player with the maximum runs using 'R'
idx = batting_df.groupby(['Date', 'Team', 'Vs'])['R'].idxmax()
highest_scorers_per_match = batting_df.loc[idx, ['Player', 'Team', 'Vs', 'Date', 'R']]
display(highest_scorers_per_match)
```

### Question 6: Do teams perform better while chasing or while batting first?

```python
# Fallback to total runs ('R') distribution per team-innings basis
total_runs_per_innings = batting_df.groupby(['Date', 'Team', 'Vs'])['R'].sum().reset_index()
display(total_runs_per_innings.describe())
```

### Question 7: Who are the top 10 bowlers based on total wickets taken, and what are their average economy rates?

```python
# Ensure Econ column is numeric
bowling_df['Econ'] = pd.to_numeric(bowling_df['Econ'], errors='coerce')

# Calculate top 10 bowlers based on wickets ('W') and aggregate economy ('Econ')
bowling_stats = bowling_df.groupby('Bowler').agg({'W': 'sum', 'Econ': 'mean'}).reset_index()
top_10_bowlers = bowling_stats.sort_values(by='W', ascending=False).head(10)
display(top_10_bowlers)
```

### Question 8: Which players have scored the most runs through boundaries (fours and sixes), and how does their strike rate compare?

```python
# Calculate boundary runs with 'R', 'B', '4s', '6s'
batting_summary = batting_df.groupby('Player').agg({'R': 'sum', 'B': 'sum', '4s': 'sum', '6s': 'sum'}).reset_index()
batting_summary['Boundary_Runs'] = (batting_summary['4s'] * 4) + (batting_summary['6s'] * 6)
batting_summary['Strike_Rate'] = (batting_summary['R'] / batting_summary['B']) * 100

top_boundary_hitters = batting_summary.sort_values(by='Boundary_Runs', ascending=False).head(10)
display(top_boundary_hitters[['Player', 'Boundary_Runs', 'Strike_Rate']])
```

### Question 9: Player Performance Score for Batsmen (Ranking Top 20)

```python
# Player Performance Score (PPS) with corrected column keys 'R' and 'B'
batting_summary = batting_df.groupby('Player').agg({'R': 'sum', 'B': 'sum', '4s': 'sum', '6s': 'sum'}).reset_index()
batting_summary['Strike_Rate'] = (batting_summary['R'] / batting_summary['B'].replace(0, 1)) * 100

batting_summary['PPS'] = (
    batting_summary['R'] * 1.0 +
    batting_summary['Strike_Rate'] * 0.5 +
    batting_summary['4s'] * 2.0 +
    batting_summary['6s'] * 4.0 -
    batting_summary['B'] * 0.1
)

top_20_batsmen = batting_summary.sort_values(by='PPS', ascending=False).head(20)
display(top_20_batsmen)
```

### Question 10: Identify top 10 all-round performers using a custom Match Impact Score

```python
# Ensure Econ and W (Wickets) columns are numeric
bowling_df['Econ'] = pd.to_numeric(bowling_df['Econ'], errors='coerce')
bowling_df['W'] = pd.to_numeric(bowling_df['W'], errors='coerce')

# Combine batting and bowling metrics using corrected column keys
all_round_batting = batting_df.groupby('Player').agg({'R': 'sum', 'B': 'sum', '4s': 'sum', '6s': 'sum', 'Win': 'sum'}).reset_index()
all_round_bowling = bowling_df.groupby('Bowler').agg({'W': 'sum', 'Econ': 'mean'}).reset_index()

all_rounders = pd.merge(all_round_batting, all_round_bowling, left_on='Player', right_on='Bowler', how='inner')
all_rounders['Batting_SR'] = (all_rounders['R'] / all_rounders['B'].replace(0, 1)) * 100

# Match Impact Score (MIS) using corrected keys
all_rounders['Match_Impact_Score'] = (
    all_rounders['R'] * 1.0 +
    all_rounders['W'] * 20.0 +
    all_rounders['Batting_SR'] * 0.2 -
    all_rounders['Econ'] * 5.0 +
    all_rounders['Win'] * 10.0
)

top_10_all_rounders = all_rounders.sort_values(by='Match_Impact_Score', ascending=False).head(10)
display(top_10_all_rounders[['Player', 'R', 'W', 'Match_Impact_Score']])
```
