```python
import pandas as pd

df_bowling = pd.read_csv('/content/DATASETS-DATASCIENCE/MatchResults refined-Bowling.csv')

df_bowling['Dots'] = pd.to_numeric(df_bowling['Dots'], errors='coerce').fillna(0).astype(int)

dot_balls_by_country = df_bowling.groupby('Team')['Dots'].sum().reset_index()
dot_balls_by_country.columns = ['Country Name', 'Total Dot Balls']

dot_balls_by_country = dot_balls_by_country.sort_values(by='Total Dot Balls', ascending=False)

display(dot_balls_by_country)
```
