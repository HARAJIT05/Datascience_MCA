```python
import pandas as pd

df = pd.read_csv('MobileSales.csv')
display(df.head())
```

# 1. Which mobile brand has the highest number of mobile models in the dataset?

```python
brand_counts = df['Brands'].value_counts()
highest_brand = brand_counts.idxmax()
highest_count = brand_counts.max()

print(f"The brand with the highest number of mobile models is '{highest_brand}' with {highest_count} entries in the dataset.")
display(brand_counts)
```

# 2. Which are the top 10 mobile models based on their selling price?

```python
# Identify columns and clean Selling Price if necessary
# Let's check the column names first in df
print("Columns:", df.columns.tolist())
# Usually columns might include 'Selling Price' or similar. We will sort by that.
selling_price_col = [col for col in df.columns if 'Selling' in col or 'price' in col.lower()][0]

top_10_models = df[['Brands', 'Models', selling_price_col]].drop_duplicates().sort_values(by=selling_price_col, ascending=False).head(10)
display(top_10_models)
```

# 3. What is the average selling price of mobiles for each brand?

```python
avg_price_brand = df.groupby('Brands')[selling_price_col].mean().sort_values(ascending=False).reset_index()
display(avg_price_brand)
```

# 4. Which mobile brand offers the highest average discount?

```python
discount_col = [col for col in df.columns if 'Discount' in col or 'discount' in col.lower()][0]
avg_discount_brand = df.groupby('Brands')[discount_col].mean().sort_values(ascending=False).reset_index()

print(f"The brand with the highest average discount is {avg_discount_brand.iloc[0]['Brands']} with {avg_discount_brand.iloc[0][discount_col]:.2f} average discount.")
display(avg_discount_brand)
```

# 5. Which mobile brands have the highest average customer rating, and how does their average selling price compare?

```python
rating_col = [col for col in df.columns if 'Rating' in col or 'rating' in col.lower()][0]

brand_stats = df.groupby('Brands').agg(
    avg_rating=(rating_col, 'mean'),
    avg_selling_price=(selling_price_col, 'mean')
).sort_values(by='avg_rating', ascending=False).reset_index()

display(brand_stats)
```

# 6. How does mobile storage capacity affect the average selling price? Can you compare mobiles with different storage options?

```python
storage_col = [col for col in df.columns if 'Storage' in col or 'storage' in col.lower() or 'ROM' in col or 'Memory' in col][0]

storage_price_comparison = df.groupby(storage_col)[selling_price_col].agg(['mean', 'count']).sort_values(by='mean', ascending=False).reset_index()
display(storage_price_comparison)
```

# 7. Which mobile models provide the highest discount amount, and what percentage of their original price is discounted?

```python
original_price_col = [col for col in df.columns if 'Original' in col or 'MRP' in col or 'Original Price' in col][0]

# Calculate discount percentage if not already a percentage, or calculate discount amount if not present
df['discount_amount'] = df[original_price_col] - df[selling_price_col]
df['discount_percentage'] = (df['discount_amount'] / df[original_price_col]) * 100

top_discounts = df[['Brands', 'Models', original_price_col, selling_price_col, 'discount_amount', 'discount_percentage']].drop_duplicates().sort_values(by='discount_amount', ascending=False).head(10)
display(top_discounts)
```

# 8. Which combination of memory and storage offers the best balance between selling price and customer rating?

```python
import pandas as pd

# Explicitly set the columns to avoid overlapping detection
memory_col = 'Memory'
storage_col = 'Storage'

# Group by memory and storage
combo_stats = df.groupby([memory_col, storage_col]).agg(
    avg_price=(selling_price_col, 'mean'),
    avg_rating=(rating_col, 'mean'),
    count=('Brands', 'count')
).reset_index()

# Filter combinations with at least 5 products for robustness
combo_stats = combo_stats[combo_stats['count'] >= 5]
combo_stats['balance_score'] = combo_stats['avg_rating'] / (combo_stats['avg_price'] + 1) * 1000

best_combos = combo_stats.sort_values(by='balance_score', ascending=False)
display(best_combos.head(10))
```

# 9. Can you identify the top 10 mobile models that provide the best value for money by considering selling price, discount, storage, memory, and customer rating? Explain how you calculated the value score.

```python
# Value Score formula:
# Value Score = (Rating * (1 + Discount_Percentage/100) * log(Storage + 1) * log(Memory + 1)) / (Selling_Price_normalized)
# Let's build a simple normalized formula:

from sklearn.preprocessing import MinMaxScaler
import numpy as np

val_df = df.copy()

# Handle missing values and convert columns to numeric
val_df[rating_col] = pd.to_numeric(val_df[rating_col], errors='coerce').fillna(3.5)
val_df[selling_price_col] = pd.to_numeric(val_df[selling_price_col], errors='coerce').fillna(val_df[selling_price_col].median())
val_df['discount_percentage'] = pd.to_numeric(val_df['discount_percentage'], errors='coerce').fillna(0)
val_df[storage_col] = pd.to_numeric(val_df[storage_col], errors='coerce').fillna(0)
val_df[memory_col] = pd.to_numeric(val_df[memory_col], errors='coerce').fillna(0)

# Normalize features between 0 and 1
scaler = MinMaxScaler()
val_df['norm_rating'] = scaler.fit_transform(val_df[[rating_col]])
val_df['norm_price'] = scaler.fit_transform(val_df[[selling_price_col]])
val_df['norm_discount'] = scaler.fit_transform(val_df[['discount_percentage']])
val_df['norm_storage'] = scaler.fit_transform(val_df[[storage_col]])
val_df['norm_memory'] = scaler.fit_transform(val_df[[memory_col]])

# Calculate Value Score: Higher ratings, higher discount, higher storage & memory, lower price
val_df['Value_Score'] = (val_df['norm_rating'] + val_df['norm_discount'] + val_df['norm_storage'] + val_df['norm_memory']) / (val_df['norm_price'] + 0.1)

top_value = val_df[['Brands', 'Models', selling_price_col, rating_col, 'discount_percentage', storage_col, memory_col, 'Value_Score']].drop_duplicates(subset=['Brands', 'Models']).sort_values(by='Value_Score', ascending=False).head(10)
display(top_value)
```

# 10. Can you analyze the factors that determine whether a mobile is classified as a "Good Product"? Which factors such as price, discount, rating, memory, or storage appear to have the strongest relationship with Good Product?

```python
# Let's check if there is a 'Good Product' column or target column in the dataset
good_prod_col = [col for col in df.columns if 'Good' in col or 'good' in col.lower()]

if good_prod_col:
    target = good_prod_col[0]
    # Calculate correlations between variables and 'Good Product'
    # Convert target to numeric 0 and 1 if it is boolean or string
    if df[target].dtype == 'object' or df[target].dtype == 'bool':
        df['Good_Product_Numeric'] = df[target].astype(str).str.lower().map({'yes': 1, 'true': 1, '1': 1, '1.0': 1, 'no': 0, 'false': 0, '0': 0, '0.0': 0})
    else:
        df['Good_Product_Numeric'] = df[target]

    numeric_cols = df[[selling_price_col, 'discount_percentage', rating_col, storage_col, memory_col, 'Good_Product_Numeric']].apply(pd.to_numeric, errors='coerce')
    correlations = numeric_cols.corr()['Good_Product_Numeric'].sort_values(ascending=False)
    print("Correlation of features with 'Good Product':")
    display(correlations)
else:
    print("No 'Good Product' column found. Let's create one based on Rating >= 4.0 and display the correlations:")
    df['Good_Product_Numeric'] = (df[rating_col] >= 4.0).astype(int)
    numeric_cols = df[[selling_price_col, 'discount_percentage', rating_col, storage_col, memory_col, 'Good_Product_Numeric']].apply(pd.to_numeric, errors='coerce')
    correlations = numeric_cols.corr()['Good_Product_Numeric'].sort_values(ascending=False)
    display(correlations)
```
